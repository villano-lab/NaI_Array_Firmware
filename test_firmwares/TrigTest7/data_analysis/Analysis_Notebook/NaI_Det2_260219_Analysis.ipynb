```python
#we may need some code in the ../python directory and/or matplotlib styles
import sys
sys.path.append('../python/')

#matplotlib for plotting
import matplotlib as mpl
from matplotlib import pyplot as plt
plt.style.use('../mplstyles/stylelib/standard.mplstyle')

#other computational libraries
import numpy as np
```

```python
import pandas as pds
```

```python
cosmics = pds.read_csv("../../cosmics_data/cosmics_NaI_Det2_260219.txt", skiprows=1, \
                         names=['Label','junk', 'BinHigh','BinLow','Counts'], \
                         sep='\s+',encoding='utf_16')
print(cosmics)
```

```python
x = np.asarray(cosmics["BinLow"])
y = np.asarray(cosmics["Counts"])
```

```python
# # creating 2D histogram of qi vs pt to visualize ER and NR bands
fig, axs = plt.subplots(1,1,figsize=(10,8))
ax1 = axs
    
ax1.steps(x, y)
    
ax1.set_xlim(0, 17000) #in pairs
ax1.set_ylim(0,1200) #

ax1.set_xlabel("Channels", fontsize=25)
ax1.set_ylabel("Counts", fontsize=25)

ax1.grid(True)
ax1.yaxis.grid(True,which='minor',linestyle='--')
#ax1.legend(loc=1,prop={'size':20})

for axis in ['top','bottom','left','right']:
  ax1.spines[axis].set_linewidth(2)

plt.tight_layout()
plt.show()
    
#change colormap here !!!
#plt.savefig(f'figures/Ba_{title}_{cmap1}.png')
```
