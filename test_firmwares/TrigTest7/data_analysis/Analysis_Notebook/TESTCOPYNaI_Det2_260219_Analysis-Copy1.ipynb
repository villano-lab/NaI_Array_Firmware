```python
#AI GENERATED - Check
#Find factors to align x axis from different voltages

#Furthest peak x location
peak_900 = 331.0 
peak_800 = 164.0
peak_700 = 89.0

#Scaling factor calculation (print if need numbers) (Can also just change to numbers if wanted)
k_800 = peak_900 / peak_800
k_700 = peak_900 / peak_700
#print(k_700)

#Actually scaling the x axis by found factors
x7_scaled = x7 * k_800
x8_scaled = x8 * k_700

fig, axs = plt.subplots(1,1,figsize=(10,8))
ax1 = axs
    
ax1.step(x6, y6, label="900V (Blue, Reference)", alpha=0.7)
ax1.step(x7_scaled, y7, label=f"800V (Orange, Scaled x{k_800:.2f})", alpha=0.7)
ax1.step(x8_scaled, y8, label=f"700V (Green, Scaled x{k_700:.2f})", alpha=0.7)

ax1.set_xlim(50, 1000) 
ax1.set_ylim(0, 30000) 

ax1.set_xlabel("Channels (Scaled to 900V Equivalent)", fontsize=25)
ax1.set_ylabel("Counts per Minute (CPM)", fontsize=25) 

ax1.grid(True)
ax1.yaxis.grid(True,which='minor',linestyle='--')
ax1.legend(loc=1,prop={'size':20})

for axis in ['top','bottom','left','right']:
  ax1.spines[axis].set_linewidth(2)

plt.tight_layout()
plt.show()
#This lined up the furthest peak, but the rest seems quite off. Is this the wrong relationship, or is that correct due to cutoff or something?
```
